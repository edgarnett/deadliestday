# Sample notebook

A small example using pandas and matplotlib.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
rng = np.random.default_rng(42)
df = pd.DataFrame({
    "day": pd.date_range("2026-01-01", periods=30, freq="D"),
    "value": rng.normal(100, 10, 30).cumsum() / 10,
})
df.head()
```

```python
df["value"].describe()
```

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(df["day"], df["value"], marker="o")
ax.set_title("Sample time series")
ax.set_xlabel("Day")
ax.set_ylabel("Value")
fig.autofmt_xdate()
plt.show()
```
